Diccionarios
=============================

Los items se almacenan y recuperar por una clave, en vez de por su posición en la estructura.

Es una tipo de agregación en el cual el nombre de los items tiene más significado que su posición.

Los diccionarios pueden reemplazar muchos de los algoritmos de búsqueda y estructuras de datos que deberías implementar manualmente en lenguajes de bajo nivel.

Características:

- Se accede a los elementos por su clave o key, no por la posición que ocupan.
- Son colecciones no ordenadas de objetos de cualquier tipo, y el tipo de los elementos que lo forman puede ser distinto.
- Poseen longitud variable y pueden anidarse.
- Son mapas mutables.
- Son tablas de referencias a objetos o hash tables.

![Table8-2](images/diccionarios-01.png)

```python
# D = {key: value}
D = {'spam': 2, 'ham': 1, 'eggs': 3}        # construir un diccionario
D
```

Technically, the ordering is pseudo-random—it’s not truly random

```python
D['spam']           # recuperar un valor por su clave
```

```python
len(D)      # Numero de entradas del diccionario
# 3
```

```python
'ham' in D      # Key membership test alternative
# True
```

```python
list(D.keys())      # Crear una lista con las claves del diccionario
# ['eggs', 'spam', 'ham']
```

Es necesario encerrar el objeto que devuelve `keys()` en el constructor `list` en Python 3.X porque `keys()` en 3.X devuelve un objeto iterable en vez de una lista física.

#### Cambiando diccionarios in place

```python
D['ham'] = ['grill', 'bake', 'fry']         # Change entry (value=list)
D
# {'eggs': 3, 'spam': 2, 'ham': ['grill', 'bake', 'fry']}
```

```python
del D['eggs']               # Delete entry
D
# {'spam': 2, 'ham': ['grill', 'bake', 'fry']}
```

Al contrario que las listas, cuando asignas una nueva clave en un diccionario (una que no haya sido asignada con anterioridad) creas una nueva entrada en el diccioanrio.

```python
D['brunch'] = 'Bacon'       # Add new entry
D
# {'brunch': 'Bacon', 'spam': 2, 'ham': ['grill', 'bake', 'fry']}
```

#### Métodos de un objeto diccionario
`values`
El método `values` devuelve todos los valores del diccionario.

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
list(D.values())                
# [3, 2, 1]
```

### `items`
`items` devuelve todas las tuplas `(key,value)`.

```python
list(D.items())
# [('eggs', 3), ('spam', 2), ('ham', 1)]
```

### `get`
Fetching una clave que no existe es normalmente un error, pero el método `get` devuelve `None` como valor por defecto o un valor pasado como por defecto si la clave no existe.

```python
D['toast']      # KeyError: 'toast'   
```

```python
D.get('spam')
# 2
```

```python
print(D.get('toast'))       # Una clave que no existe
# None
```

```python
D.get('toast', 88)          # proporcionando un valor por defecto
# 88
```

### `update`
El metodo `update` une las claves y valores de un diccionaroi en otro, sobreescribiendo valores de la misma clave si hay conflicto.

```python
# D = {'eggs': 3, 'spam': 2, 'ham': 1}
D2 = {'toast':4, 'muffin':5, 'ham': 10}
D.update(D2)
D
# {'eggs': 3, 'muffin': 5, 'toast': 4, 'spam': 2, 'ham': 10}
```

### `pop`
`pop` elimina una clave de un diccionario y devuelve el valor asociado a la clave.

```python
D.pop('muffin')
# 5
D.pop('toast')
# 4
D
# {'eggs': 3, 'spam': 2, 'ham': 10}
```

### Vistas diccionario y sets
La vista objeto devuelta por el método `keys` es como un `set` y soporta las operaciones con conjuntos, como la intersección y la unión.

```python
K, V
```

```python
K | {'x': 4}            # La vista `keys` es como un `set` 
```

La vista `values` no es como un `set`.

```python
V & {'x': 4}            # TypeError: unsupported operand type(s) for &: 'dict_values' and 'dict'
```

```python
V & {'x': 4}.values()   # TypeError: unsupported operand type(s) for &: 'dict_values' and 'dict_values'
```

### Section 19.2: Avoiding KeyError Exceptions
Una de las trampas más comunes es el uso de diccionarios cuando estos tienen acceso a una key no existente. Esto resulta en un KeyError.

```python
mydict = {}
mydict['not there']
```

Una de las formas de evitar un error de key es usar el método `dict.get`, el cuál te permite especificar un valor por defecto para devolver en caso de una clave ausente.

```python
value = mydict.get(key, default_value)
```

Esto devuelve `mydict[key]` si existe; de ​​lo contrario, devuelve `default_value`. Ten en cuenta que esto no añade `key` a `mydict`. Por tanto, si deseas conservar ese par clave-valor, deberías utilizar `mydict.setdefault(key, default_value)`, que sí almacena dicho par clave-valor.

```python
mydict = {}
print(mydict)
print(mydict.get("foo", "bar"))
print(mydict)
print(mydict.setdefault("foo", "bar"))
print(mydict)
```

### Section 19.3: Iterating Over a Dictionary
Si utilizas un diccionario como iterador (por ejemplo, en una sentencia `for`), este recorre las claves del diccionario. Por ejemplo:

```python
d = {'a': 1, 'b': 2, 'c':3}
for key in d:
    print(key, d[key])
# c 3
# b 2
# a 1
```

También es Verdadero cuando se usa en una comprensión

```python
print([key for key in d])
# ['c', 'b', 'a']
```

```python
for key, value in d.items():
    print(key, value)
# 3
# 2
# 1
```

### Section 19.7: Accessing values of a dictionary


```python
dictionary = {"Hello": 1234, "World": 5678}
print(dictionary["Hello"])
```

En este ejemplo, la cadena «Hello» recibe el nombre de «clave» (key). Se utiliza para buscar un valor en el diccionario colocando dicha clave entre corchetes.
El número 1234 aparece después de los dos puntos correspondientes en la definición del diccionario. A esto se le denomina el «valor» al que «Hello» está asociado (o mapeado) dentro de este diccionario.
Si se intenta buscar un valor mediante una clave que no existe, se producirá una excepción de tipo `KeyError`, lo cual detendrá la ejecución del programa si dicha excepción no es capturada. Si deseamos acceder a un valor sin correr el riesgo de que surja un `KeyError`, podemos emplear el método `.get()` del diccionario. Por defecto, si la clave no existe, el método devolverá `None`. Asimismo, es posible proporcionar un segundo argumento para que, en caso de que la búsqueda falle, el método devuelva ese valor en lugar de `None`.

```python
w = dictionary.get("whatever")
x = dictionary.get("whatever", "nuh-uh")
```

En este ejemplo, w tomará el valor `None` y x tomará el valor "nuh-uh".

###  Section 19.8: Creating a dictionary
Reglas para crear un diccionario
- Todas las claves deben ser únicas (el resto serán anuladas)
- Todas las claves deben ser hashable (un coódigo hash que no cambia)
- No hay un orden en particular para la claves

```python
# Creating and populating it with values
stock = {'eggs': 5, 'milk': 2}
stock
```

```python
# Or creating an empty dictionary
dictionary = {}
```

```python
# And populating it after
dictionary['eggs'] = 5
dictionary['milk'] = 2
dictionary
```

```python
# Values can also be lists
mydict = {'a': [1, 2, 3], 'b': ['one', 'two', 'three']}
mydict
```

```python
# Use list.append() method to add new elements to the values list
mydict['a'].append(4) # => {'a': [1, 2, 3, 4], 'b': ['one', 'two', 'three']}
mydict['b'].append('four') # => {'a': [1, 2, 3, 4], 'b': ['one', 'two', 'three', 'four']}
mydict
```

```python
# We can also create a dictionary using a list of two-items tuples
iterable = [('eggs', 5), ('milk', 2)]
dictionary = dict(iterable)
dictionary
```

```python
# Or using keyword argument:
dictionary = dict(eggs=5, milk=2)
dictionary
```

```python
# Another way will be to use the dict.fromkeys:
dictionary = dict.fromkeys((milk, eggs)) # => {'milk': None, 'eggs': None}
dictionary = dict.fromkeys((milk, eggs), (2, 5)) # => {'milk': 2, 'eggs': 5}
```
