#### Paso de objetos mutables:

```python
def changer(a, b):          # Arguments assigned references to objects
                a = 2                       # Changes local name's vaslue only
                b[0] = 'spam'               # Changes shared object in place

x = 1
l = [1, 2]
changer(x, l)               # Caller: Pass inmutable and mutable objects
print("l = ", l)            # ['spam', 2]   # x is unchanged, l is different!
assert x == 1, "x es 1"     # assert comprube asi es correcto el parámetro indicado, True or False.
```

#### Cómo evitar que una rutina moifique los elementos mutables:

##### - Pasar una copia

```python
l = [1, 2]
changer(x, l[:])        # Pass a copy, so our 'l' does not change

print("l = ", l)
```

##### - Utilizar invariantes (en las estructuras de datos de entrada)

```python
l = [1, 2]
def changer(a, b):
    b = b[:]                # Copy input list so we don't impact caller
    a = 2
    b[0] = 'spam'           # Changes our list copy only
    
    print(a, b)
    
changer(x, l)
assert l == [1, 2], "l es una invariante"
print("l = ", l)
```

##### - Convertir el objeto mutable a un objeto imutable:

```python
l = [1, 2]
def changer(a, b):
                a = 2
                b[0] = 'spam'           #TypeError: 'tuple' object does not support item assignment
                
changer(x, tuple(l))                 # Pass a tuple, so change are errors

print("l = ", l)
assert l == [1, 2], "l es una invariante"
```
