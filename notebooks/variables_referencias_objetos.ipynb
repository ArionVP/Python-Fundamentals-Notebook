### OBJETOS MUTABLES

##### Listas

```python
L = [2, 3 ,4]           # Es un objeto mutable
M = L                   # Se crea una referencia para el mismo objeto
L[0] = 24               # Cambia el valor del primer dígito de la lista

print("L = ", L)        # [24, 3, 4]
print("M = ", M)        # [24, 3, 4]    
```

Copia de objetos en vez de referencia:

```python
L = [2, 3, 4]
M = L[:]                # Se crea una copia de L (o list(L), copy.copy(L), etc.)
L[0] = 24
print("L = ", L)        # [24, 3, 4]
print("M = ", M)        # [2, 3, 4] M no cambia el valor
```

#### Shared References and Equality
- Operador `==` igualdad de valores.
- Operador `is` identidad de objetos.

```python
L = [1, 2, 3]
M = [1, 2, 3]       # M y L referencian su mismo valor a diferentes objetos.

print(
    "L == M",
    L == M          # True : Mismos valores
)

print(
    "L is M",       # False : el operador identidad de objetos: compara las referencias (los punteros)
    L is M
)
```

##### CACHE
Por algunos motivos los small integers y strings son almacenados en cache y reutilizados, así mismo, nos dice la referencia de ellos hacia un mismo objeto único.

```python
X = 42 
Y = 42          # Debería de ser 2 objetos diferentes

X == Y          # True
X is Y          # True: Mismo objeto de todos modos: caching at work !!
```

Averiguar el nḿero de referencias a un objeto:

```python
import sys 
sys.getrefcount(1)          # número de punteros compartidos a la pieza de memoria.
```
